```python
a0d=np.array(40);
print(f'array: {a0d}')
print(f'shape: {a0d.shape}')
print(f'dimension: {a0d.ndim}')
print(f'size: {a0d.size}')
print(f'type: {a0d.dtype}')
```

```python
a2d=np.array([[1,2,3],[4,5,6]])
print(f'array: {a2d}')
print(f'shape: {a2d.shape}')
print(f'dimension: {a2d.ndim}')
print(f'size: {a2d.size}')
print(f'type: {a2d.dtype}')
```

```python
a3d=np.array([[1,2,3],[7,8,9],[6,5,4],[3,9,2]])
print(f'array: {a3d}')
print(f'shape: {a3d.shape}')
print(f'dimension: {a3d.ndim}')
print(f'size: {a3d.size}')
print(f'type: {a3d.dtype}')
print(f'total size: {a3d.nbytes}')
```

```python
print(a3d[0])
print(a3d[1])
print(a3d[2])
print(a3d[3])
```

```python
print(f'Second row first element: {a3d[1,0]}')
print(f'Second row second element: {a3d[1,1]}')
print(f'Second row third element: {a3d[1,2]}')
```

```python
print(f'Second column first element: {a3d[0,1]}')
print(f'Second column second element: {a3d[1,1]}')
print(f'Second column third element: {a3d[2,1]}')
```

```python
print(f'row: {a3d[0,:]}')
print(f'row: {a3d[1,:]}')
print(f'row: {a3d[2,:]}')
```

```python
marks1=np.array([34,65,87,12,54,76])
print(f'first four: {marks1[:4]}')
print(f'last two: {marks1[-2:]}')
print(f'every other: {marks1[::2]}')
print(f'reverse: {marks1[::-1]}')
print(f'reverse every other: {marks1[::-2]}')
print(f'alternate: {marks1[::2]}')
```
